## Inicio

**Ubicación:** terminal de Visual Studio Code en la carpeta `Orbit_Studio_alumno`.

```bash
npm ci
npm run dev
```

Abre la dirección de **Local**. Conserva la terminal abierta. Los comandos sirven en Mac, Windows y Linux con Node.js y npm instalados.

## 2. Cabeza

**Archivo:** `src/personaje.js`

Conserva el grupo existente y agrega su pieza. Evita duplicar la declaración del grupo.

```javascript
const cabeza = grupo('cabeza', [0, 2.4, 0]);
pieza('cabezaBase', new THREE.BoxGeometry(0.8, 0.7, 0.7), materiales.claro, [0, 0.38, 0], cabeza);
```

## 3. Un brazo

**Archivo:** `src/personaje.js`

Actualiza el grupo existente y agrega su pieza. Y local negativa extiende el brazo hacia abajo.

```javascript
const brazoIzquierdo = grupo('brazoIzquierdo', [-0.92, 2.08, 0]);
pieza('brazoIzq', new THREE.BoxGeometry(0.3, 0.8, 0.35), materiales.oscuro, [0, -0.4, 0], brazoIzquierdo);
```

## 4. Ojos

**Archivo:** `src/personaje.js`

Agrega las piezas después de crear el grupo cabeza. X separa los ojos y Z positiva los lleva al frente.

```javascript
pieza('ojoIzquierdo', new THREE.BoxGeometry(0.12, 0.16, 0.06), materiales.oscuro, [-0.2, 0.45, 0.36], cabeza);
pieza('ojoDerecho', new THREE.BoxGeometry(0.12, 0.16, 0.06), materiales.oscuro, [0.2, 0.45, 0.36], cabeza);
```
